```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt('/users/wangwenfei/desktop/python_decal_fa25/bu/homework11/astro_image.csv')
```

```python
image = data.reshape((1600, 1600))
```

```python
plt.figure(figsize=(10, 8))
plt.imshow(image, cmap='viridis')
plt.colorbar(label='Intensity')
plt.title('Astronomical Image (1600×1600)')
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
center = 800
zoom = 200
zoom1 = image[center-zoom:center+zoom, center-zoom:center+zoom]
ax1.imshow(zoom1, cmap='viridis')
ax1.set_title('Zoom: Center Region (600:1000, 600:1000)')
max_pos = np.unravel_index(np.argmax(image), (1600, 1600))
zoom2 = image[max_pos[0]-zoom:max_pos[0]+zoom, max_pos[1]-zoom:max_pos[1]+zoom]
ax2.imshow(zoom2, cmap='viridis')
ax2.set_title(f'Zoom: Brightest Feature\nat ({max_pos[1]}, {max_pos[0]})')
plt.tight_layout()
plt.show()
```
